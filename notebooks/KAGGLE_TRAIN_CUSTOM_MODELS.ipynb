# Part 3: Generate Training Data

```python
# Sample Konkani sentences
konkani_sentences = [
    "हांव घरा वता",
    "तुजें नांव कितें?",
    "हांव खूश आसा",
    # Add more...
]

print(f"Sample sentences: {len(konkani_sentences)}")
```

```python
# Generate translations
from googletrans import Translator
from tqdm.auto import tqdm

translator = Translator()
translation_pairs = []

for text in tqdm(konkani_sentences):
    try:
        result = translator.translate(text, src='hi', dest='en')
        translation_pairs.append({'konkani': text, 'english': result.text})
    except:
        continue

print(f"✅ Generated {len(translation_pairs)} pairs")
```
